```python
sampleA = [10, 50, 100, 150]
sampleB = [150, 100, 50, 10]
```

```python
import numpy as np 

sampleA = np.array(sampleA)
sampleB = np.array(sampleB) 

sample_sum = abs(np.sum(sampleA - sampleB))
eq = sample_sum/total 
print(eq)
```
